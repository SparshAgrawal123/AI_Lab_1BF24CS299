```python

from itertools import product

# Knowledge Base using propositional logic
# Example:
# P: It is raining
# Q: The ground is wet
# R: The road is slippery

# KB:
# 1. P -> Q
# 2. Q -> R
# 3. P

def implies(a, b):
    return (not a) or b

def knowledge_base(P, Q, R):
    rule1 = implies(P, Q)  # P -> Q
    rule2 = implies(Q, R)  # Q -> R
    fact1 = P              # P is true

    return rule1 and rule2 and fact1

# Query to check
# Query: R (The road is slippery)

def query(P, Q, R):
    return R

# Generate all possible truth assignments
symbols = ['P', 'Q', 'R']

entailment = True
models = 0

print("TRUTH TABLE")
print("P\tQ\tR\tKB\tQuery")

for values in product([True, False], repeat=3):
    P, Q, R = values

    kb = knowledge_base(P, Q, R)
    q = query(P, Q, R)

    print(P, Q, R, kb, q, sep="\t")

    # Check whether KB is true
    if kb:
        models += 1

        # If KB is true but query is false,
        # entailment does not hold
        if not q:
            entailment = False

print("\nRESULT:")

if entailment and models > 0:
    print("KB entails the Query (KB |= Query)")
else:
    print("KB does NOT entail the Query")
```
