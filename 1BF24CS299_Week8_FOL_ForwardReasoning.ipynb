```python

# Forward Chaining in First Order Logic

# Knowledge Base
facts = [
    ("Human", "Socrates")
]

# Rules in the form:
# (condition, conclusion)

rules = [
    ("Human", "Mortal"),
    ("Mortal", "Dies")
]


def forward_chaining(facts, rules, query):
    # Store all known facts
    known = list(facts)

    print("Initial Facts:")
    for fact in known:
        print(fact)

    print("\nForward Reasoning:")

    while True:
        new_fact_added = False

        for rule in rules:
            condition, conclusion = rule

            # Apply rule to every known fact
            for fact in known[:]:

                if fact[0] == condition:
                    new_fact = (conclusion, fact[1])

                    if new_fact not in known:
                        known.append(new_fact)

                        print(
                            condition, fact[1],
                            "->",
                            conclusion, fact[1]
                        )

                        new_fact_added = True

        # Stop if no new facts are generated
        if not new_fact_added:
            break

    print("\nFinal Knowledge Base:")
    for fact in known:
        print(fact)

    # Check whether query exists in knowledge base
    print("\nQuery:", query)

    if query in known:
        print("Query is PROVED using Forward Chaining!")
        return True
    else:
        print("Query CANNOT be proved.")
        return False


# Given Query
query = ("Mortal", "Socrates")

# Execute forward chaining
forward_chaining(facts, rules, query)
```
