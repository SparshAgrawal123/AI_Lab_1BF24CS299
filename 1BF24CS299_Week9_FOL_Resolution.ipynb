```python

# First Order Logic Resolution

# Knowledge Base in CNF (Conjunctive Normal Form)
# 1. All humans are mortal:
#    NOT Human(x) OR Mortal(x)
#
# 2. Socrates is a human:
#    Human(Socrates)
#
# Query: Mortal(Socrates)

# Represent literals as:
# (predicate, argument, negated)

def negate(literal):
    predicate, argument, negated = literal
    return (predicate, argument, not negated)


def substitute(literal, subst):
    predicate, argument, negated = literal

    while argument in subst:
        argument = subst[argument]

    return (predicate, argument, negated)


def unify(x, y, subst):
    x = substitute(x, subst)
    y = substitute(y, subst)

    if x == y:
        return subst

    if x[0] != y[0] or x[2] != y[2]:
        return None

    if x[1] == y[1]:
        return subst

    # Variables are represented by lowercase letters
    if x[1].islower() and len(x[1]) == 1:
        subst[x[1]] = y[1]
        return subst

    if y[1].islower() and len(y[1]) == 1:
        subst[y[1]] = x[1]
        return subst

    return None


def resolve(c1, c2):
    for l1 in c1:
        for l2 in c2:

            # Check complementary literals
            if l1[0] == l2[0] and l1[2] != l2[2]:

                subst = unify(l1, l2, {})

                if subst is not None:

                    new_clause = []

                    for literal in c1:
                        if literal != l1:
                            new_clause.append(
                                substitute(literal, subst)
                            )

                    for literal in c2:
                        if literal != l2:
                            new_clause.append(
                                substitute(literal, subst)
                            )

                    return tuple(new_clause)

    return None


def resolution(kb, query):

    # Negate the query and add it to KB
    clauses = kb + [(negate(query),)]

    print("Initial Clauses:")
    for clause in clauses:
        print(clause)

    print("\nResolution Steps:")

    while True:

        new_clauses = []

        for i in range(len(clauses)):
            for j in range(i + 1, len(clauses)):

                result = resolve(clauses[i], clauses[j])

                if result is not None:

                    print(clauses[i], "AND", clauses[j])
                    print("=>", result)

                    # Empty clause means contradiction
                    if len(result) == 0:
                        print("\nEmpty clause obtained!")
                        print("Query is PROVED by Resolution.")
                        return True

                    if result not in clauses:
                        new_clauses.append(result)

        if not new_clauses:
            print("\nNo new clauses generated.")
            print("Query cannot be proved.")
            return False

        clauses.extend(new_clauses)


# Knowledge Base in CNF
kb = [
    (
        ("Human", "x", True),
        ("Mortal", "x", False)
    ),
    (
        ("Human", "Socrates", False),
    )
]

# Given query: Mortal(Socrates)
query = ("Mortal", "Socrates", False)

# Execute resolution
resolution(kb, query)
```
